## Question 1: Weekly Sales

```python
import numpy as np 
sales = np.array([1200, 1500, 1800, 1700, 1600, 2000, 2100])

# Find total sales.
Total_sales = np.sum(sales)
print(Total_sales)

# Find average sales.
average_sales = np.average(sales)
print(average_sales)

# Find highest sales.
highest_sales = np.max(sales)
print(highest_sales)

# Find lowest sales.
lowest_sales = np.min(sales)
print(lowest_sales)

# Find the day with maximum sales.
max_day_sales = np.argmax(sales) # argmax --> returns the index position of the Max value
print(max_day_sales)

# Find the day with minimum sales.
min_day_sales = np.argmin(sales) # argmin --> returns the index position of the Min value
print(min_day_sales)
```

# Question 2: Student Marks

```python
import numpy as np
marks = np.array([67,45,89,90,78,56,99,100,34,81])

# Count students scoring above 80.
index = np.where(marks>80)
print(marks[index])

# Count failed students (<40).
index = np.where(marks<40)
print(marks[index])

# Find average marks.
average_marks = np.average(marks)
print(average_marks)

# Find top 3 marks.
slice = marks[0:3]
sort = np.sort(slice)
print(sort) 

# Sort the marks.
result = np.sort(marks)
print(result)
```

# Question 4: Product Ratings

```python
import numpy as np
ratings = np.array([5,4,3,5,2,1,4,5,3,2,4,5])

# Count 5-star ratings.
count1 = np.where(ratings==5)
print(len(ratings[count1]))

# Count ratings below 3.
count2 = np.where(ratings == 3)
print(len(ratings[count2]))

# Percentage of satisfied customers (rating ≥4).
count3 = np.sum(ratings >= 4)
percentage = (count3 / len(ratings)) * 100
print("Percentage:", percentage)
```

# Question 7: Stock Market 

```python
import numpy as np 
stock = np.array([210,215,220,208,230,240,238,250])

# Daily profit/loss.
daily_changes = np.diff(stock)
print(daily_changes)

# Highest jump.
Highest_jump = np.max(daily_changes)
print(Highest_jump)

# Lowest jump 
Lowest_jump = np.min(daily_changes)
print(Lowest_jump)

# Percentage Increase from Day 1
Percentage_increase = ((stock[-1] - stock[0]) / stock[0]) * 100
print(Percentage_increase)
```

# Question 8: Real Data Science Practice

```python
import numpy as np 
customers = np.array([
[101,24,4500],
[102,31,8000],              
[103,20,2500],      # Customer ID   }
[104,42,12000],     # Age           }---> Column names
[105,36,6000],      # Amount Spent  }
[106,28,15000]
])

# Average customer age.
average_age = np.mean(customers[:,1])
print("Average Customer Age:", average_age)

# Average purchase amount.
average_purchase = np.mean(customers[:,2])
print("Average Purchase Amount:", average_purchase)

# Customer spending the most
index = np.argmax(customers[:,2])
print("Customer ID:", customers[index,0])
print("Age:", customers[index,1])
print("Amount Spent:", customers[index,2])

# Customers spending above ₹7000.
index = np.where(customers[:,2] > 7000)
print(customers[index])

# Customers Younger Than 30.
result = customers[customers[:,1] < 30]
for customer in result:                     # ---> In loops customer take single rows at a time  
    print("Customer ID:", customer[0])
    print("Age:", customer[1])
    print("Amount:", customer[2])
    print()

# Oldest Customer 
index = np.argmax(customers[:,1])
print("Customer ID:", customers[index,0])
print("Age:", customers[index,1])
print("Amount:", customers[index,2])

# Youngest Customer
index = np.argmin(customers[:,1])
print("Customer ID:", customers[index,0])
print("Age:", customers[index,1])
print("Amount:", customers[index,2])
```
