```python
from matplotlib import pyplot as plt
import pandas as pd

data = pd.read_csv("data.csv")

age_x = data["Age"]
py_dev_y = data["Python"]
dev_y = data["All_Devs"]

plt.figure(figsize=(14, 8), dpi=150)

plt.plot(
    age_x,
    py_dev_y,
    color="#306998",      
    linewidth=3,
    marker="o",
    markersize=5,
    label="Python Developers"
)

plt.plot(
    age_x,
    dev_y,
    color="#ff7f0e",      
    linewidth=3,
    linestyle="--",
    marker="s",
    markersize=5,
    label="All Developers"
)

plt.fill_between(
    age_x,
    py_dev_y,
    dev_y,
    where=(py_dev_y > dev_y),
    interpolate=True,
    color="green",
    alpha=0.25,
    label="Python Above Average"
)

plt.fill_between(
    age_x,
    py_dev_y,
    dev_y,
    where=(py_dev_y <= dev_y),
    interpolate=True,
    color="red",
    alpha=0.20,
    label="Python Below Average"
)


plt.title("Median Salary by Age",fontsize=20,fontweight="bold",pad=20)
plt.xlabel("Age", fontsize=14)
plt.ylabel("Median Salary (USD)", fontsize=14)

plt.grid( linestyle="--", linewidth=0.6, alpha=0.6 )
plt.legend(loc="upper left",)
plt.tight_layout()
plt.show()
```
