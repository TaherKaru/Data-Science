```python
from matplotlib import pyplot as plt 
import pandas as pd 
import numpy as np 

plt.figure(figsize=(14,8))

data = pd.read_csv("2019-05-31-data.csv")
count = data['view_count']
like = data["likes"]
ratio = data['ratio']

color = np.random.randint(1,10,len(like))

plt.scatter(count, like, edgecolors="black", alpha=0.7,c=color )

cbar = plt.colorbar()
cbar.set_label('Like/Dislike Ratio')

plt.xscale('log')
plt.yscale('log')

plt.title('Trending YouTube Videos')
plt.xlabel('View Count')
plt.ylabel('Total Likes')

plt.tight_layout()
plt.grid()
plt.show()
```
