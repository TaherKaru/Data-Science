```python
from matplotlib import pyplot as plt
import pandas as pd

data = pd.read_csv("data.csv")
age_x = data["Age"]
All_Devs_y = data["All_Devs"]
Python_y = data["Python"]
JavaScript_y = data["JavaScript"]

fig, (axix1,axix2) = plt.subplots(nrows = 2 , ncols = 1, figsize=(12, 8))

axix1.plot(age_x, All_Devs_y, color = "red", label = "All developers" )
axix2.plot(age_x, JavaScript_y, color = "Green", linestyle = "--", label = "Javascript dev")
axix2.plot(age_x, Python_y, color = "blue", label = "Javascript dev")

fig.suptitle('Median Salary (USD) by Age')
axix1.legend()
axix2.legend()

axix1.grid()
axix2.grid()
plt.tight_layout()
plt.show()
```
