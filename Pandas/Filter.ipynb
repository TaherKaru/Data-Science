# Filter in Pandas

```python
import pandas as pd
df = pd.read_csv("ai_student_impact_dataset (1).csv")
df
```

```python
cgpa = df['Major_Category'] == "Business"
df[cgpa]
```

```python
filt2 = df["Student_ID"] >= 149991
df.loc[filt2,"Year_of_Study"]
```

```python
skill = ['Intermediate' , "Advanced"]
data = df["Prompt_Engineering_Skill"].isin(skill)
display(df.loc[data, ["Student_ID"]])
```

# Logical operator

```python
filt2 = (df["Student_ID"] >= 145731 ) & (df['Major_Category'] == "Business")
df.loc[filt2,["Year_of_Study","Primary_Use_Case"]]
```

```python
filt2 = (df["Student_ID"] >= 145731 ) & (df['Major_Category'] == "Business") | (df['Prompt_Engineering_Skill'] == "Advanced")
df.loc[filt2,["Year_of_Study","Primary_Use_Case"]]
```

```python
filt2 = ~ ((df["Student_ID"] >= 145731 ) & (df['Major_Category'] == "Business") | (df['Prompt_Engineering_Skill'] == "Advanced"))
df.loc[filt2,["Year_of_Study","Primary_Use_Case"]]
```
