# Handling the missing values


```python
import pandas as pd 
import numpy as np 
dataset = {
    'First' : ['corey' , 'Jane' , 'John' , 'Chris' , np.nan , None , np.nan],
    'Last' : ['schafer' , 'Doe' , 'Doe' , 'schafer' , np.nan ,np.nan , 'Missing'],
    'age' : ['33','55','63','36',None, None,'Missing']
}

df = pd.DataFrame(dataset)
display(df)
```

```python
df.dropna()  # --> drop columns that contain the missing values like NaN or None
```

```python
df.dropna(axis="index" , how="any") # --> set to 'any' >> drops if any value missing
```

```python
df.dropna(axis="index" , how="all") # --> set to 'all' >> drops every row or column having all missing value
```

```python
df.dropna(axis='index' , how='any' , subset=['Last']) # --> subset >> allow to target specific column
```

# Filling missing values

```python
df.isna() # --> returns boolean showing walue treated as missing
```

```python
df.fillna('Missing') # --> replace NaN with any string value 
```
