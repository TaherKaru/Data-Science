# Working with the columns


```python
import pandas as pd
data = {
    'First':['convey','John','Jane'],
    'Last':['schafer','Doe','Doe'],
    'email':['corveyscafer@gmail.com','Johndoe@gmail.com','Janedoe@gmail.com']
}
df = pd.DataFrame(data)         # --> creating our own Dataframe
display(df)
```

```python
df["Full_name"] = df['First'] + ' ' + df['Last']  # --> creating the new column
display(df)
```

```python
df.drop(columns=['First','Last']) # -->  deleting the column First and Last 
```

```python
df['Full_name'].str.split(' ',expand=True) # --> spliting on the basis of the space given in inverted comma 
```

# Working on the rows

```python
df.loc[len(df)] = ['Owais','Shk','owais123@gmail.com', 'Owais Shk' ] # --> To add new row 
display(df)
```

```python
df2 = {
    'First' : ['Tony','Steve'],
    'Last' : ['Stark','Rogers'],
    'email' : ['Ironman@gmail.com','CaptainAmerica@gmail.com'],
    'Full_name' : ['Tony Stark' , 'Steve Rogers']
}
df2 = pd.DataFrame(df2) # --> making dataframe df2
display(df2)
```

```python
df3 = pd.concat([df,df2] ,ignore_index=True) # --> Concatinating the two dataframes (df and df2) in df3
display(df3)
```

```python
df.drop(index=0) # --> droping the row using index
```

```python
df = df.drop(index = df[df['Last'] == "Doe"].index )  # --> deletes the person with last name "Doe"
display(df)
```
