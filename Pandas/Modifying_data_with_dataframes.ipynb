# Updating columns names

```python
import pandas as pd
df = pd.read_csv("ai_student_impact_dataset (1).csv", index_col='Student_ID')
df
```

```python
df.columns=['experience'] # --> Gives error; use this when u want to change all column names otherwise use rename() method
```

```python
df.rename(columns = {
    'Year_of_Study': 'Experience',
    'Major_Category' : 'Field_of_study',
    'Weekly_GenAI_Hours' : 'Weekly_GenAI_Usage'
})                                                  # --> Column name changed
```

```python
df.columns =[x.upper() for x in df.columns] # --> Applying function on the columns
df
```
