# Sorting 


```python
import pandas as pd 
df = pd.read_csv("ai_student_impact_dataset (1).csv" , index_col="Student_ID")
df.head()
```

```python
df.sort_values(by='Post_Semester_GPA', ascending=False) # --> Basic Sorting
```

```python
df.sort_values(by=['Pre_Semester_GPA','Post_Semester_GPA'], ascending=[False,False]) # --> Multiple sorting 
```

```python
df.sort_index(ascending=False) # --> Sorting the index
```

```python
df['Anxiety_Level_During_Exams'].nlargest(1000).tail() # --> Sorting largest values in series
```

```python
df.nlargest(10,'Perceived_AI_Dependency').tail(10) # --> Sorting on Dataframe
```
