# Basic Aggregate Function


```python
import pandas as pd
df = pd.read_csv("ai_student_impact_dataset (1).csv", index_col='Student_ID')
df
```

```python
df['Pre_Semester_GPA'].median() # --> gives middle value of the dataset 
```

```python
df.describe() # --> gives broad statistical overview of the data 
```

```python
df['Pre_Semester_GPA'].count() # --> gives count of Non-missing / non-NaN rows for specific column
```

```python
df['Major_Category'].value_counts()  # --> Tallies unique response in the column
```

# Grouping the data 

```python
major_group = df.groupby('Major_Category') # --> creating a group by specifying the column
major_group.get_group('Humanities') # --> getting the group according the values
display(major_group.get_group('Humanities').value_counts()) # --> applying function on the group 
```

# Advance grouping techniques

```python
import numpy as np 
df['Anxiety_Level_During_Exams'].agg([np.mean, np.median]) # --> use to run several function 
```

```python
major_group['Major_Category'].apply(lambda x : x.str.contains('Humanities')).sum()
```
